# HEAL-CITY — Phase 1: Data Audit
This notebook loads the raw worksheets from `data smart city.xlsx` and audits the dimensions, columns, missing values, duplicates, and data types.

```python
import pandas as pd
import os

excel_path = "../data/raw/data smart city.xlsx"
excel = pd.ExcelFile(excel_path)
print("Sheets in raw workbook:", excel.sheet_names)
```

## Dataframe Auditing Function
We define a function to audit each sheet as specified in the preprocessing guidelines.

```python
def audit_dataframe(df, name):
    print("=" * 60)
    print(f"Sheet: {name}")
    print("=" * 60)
    print("Shape:", df.shape)
    print("\nColumns:", df.columns.tolist())
    print("\nMissing values per column:")
    print(df.isna().sum())
    print(f"\nDuplicate rows: {df.duplicated().sum()}")
    print("\nData types:")
    print(df.dtypes)
    print("\nFirst 2 rows:")
    display(df.head(2))
    print()

# Audit all sheets
for sheet in excel.sheet_names:
    # For kunjungan puskesmas, let's load with defaults to audit the raw layout
    df_raw = pd.read_excel(excel_path, sheet_name=sheet)
    audit_dataframe(df_raw, sheet)
```
